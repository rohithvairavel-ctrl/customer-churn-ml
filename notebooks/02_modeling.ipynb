# 02 - Modeling: Churn Prediction

Compare Logistic Regression, Random Forest, and XGBoost.
Full run: `python scripts/train.py`


```python
import sys, json
from pathlib import Path
import pandas as pd
PROJECT_ROOT = Path('..').resolve()
sys.path.insert(0, str(PROJECT_ROOT))
from src.data import load_raw, get_feature_target, feature_columns
from src.train import split_data, train_all, find_best_threshold
from src.evaluate import evaluate_models
X, y = get_feature_target(load_raw())
numeric, categorical = feature_columns(X)
X_train, X_test, y_train, y_test = split_data(X, y)
fitted = train_all(X_train, y_train, numeric, categorical)
thresholds = {n: find_best_threshold(y_train, p.predict_proba(X_train)[:,1]) for n,p in fitted.items()}
results = evaluate_models(fitted, X_test, y_test, thresholds=thresholds)
pd.DataFrame([{**{'model':n}, **{k:v for k,v in r.items() if k in ['roc_auc','pr_auc','f1','precision','recall','threshold']}} for n,r in results.items()]).sort_values('roc_auc', ascending=False)
```
