# 01 - Exploratory Data Analysis: Telco Customer Churn

**Dataset:** IBM Sample Telco Customer Churn (7,043 customers)
Run after: `python scripts/download_data.py`


```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
PROJECT_ROOT = Path('..').resolve()
sys.path.insert(0, str(PROJECT_ROOT))
from src.data import load_raw, clean_dataframe, TARGET_COL
df = load_raw()
print(df.shape, df['Churn'].value_counts(normalize=True))
df.head()
```

```python
clean = clean_dataframe(df)
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
sns.histplot(data=clean, x='tenure', hue=TARGET_COL, bins=24, ax=axes[0])
sns.boxplot(data=clean, x=TARGET_COL, y='MonthlyCharges', ax=axes[1])
sns.countplot(data=df, x='Contract', hue='Churn', ax=axes[2])
plt.tight_layout(); plt.show()
```

## Takeaways
- Churn ~26.5% (imbalanced)
- Month-to-month and fiber correlate with higher churn
- Blank TotalCharges for new customers -> impute 0
